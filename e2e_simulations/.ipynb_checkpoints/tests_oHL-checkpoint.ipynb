```python
import numpy as np
import healpy as hp
import pymaster as nmt
import matplotlib.pyplot as plt
```

```python
# Inputs

nside = 64
lmax = 2*nside-1
Nlbin = 10
fsky = '0.5_402_baseline'
scale = 10
complexity = 'baseline' # Should be 'baseline', 'medium_complexity' or 'high_complexity'
order = '1bts'
fix = 1
adaptative = True
cov_type = 'Nmt-fg'
kw = ''

if complexity == 'baseline':
    dusttype, synctype = 'b', 'b'
elif complexity == 'medium_complexity':
    dusttype, synctype = 'm', 'm'
elif complexity == 'high_complexity':
    dusttype, synctype = 'h', 'h'

if adaptative:
    kw += '_adaptative'

name = 'moments_%s_%s_%s_o%s%s' % (complexity, fsky, cov_type, order, kw) # Name of the config
path = '../BmodeLike/inputs/%s' % (name)
```

```python
# Load results

results = np.load('../best_fits/results_d%ss%s_%s_scale%s_Nlbin%s_%s_gaussbeam_bandpass_ds_o%s%s_analytical.npy'
               % (dusttype, synctype, fsky, scale, Nlbin, cov_type, order, kw), allow_pickle=True).item()
```

```python
# Compute CMB BB spectrum for each simulation

b = nmt.NmtBin.from_lmax_linear(lmax=lmax, nlb=1, is_Dell=False)
leff = b.get_effective_ells()
Nbins = len(leff)
lmins = [int(b.get_ell_min(i)) for i in range(Nbins)]
lmaxs = [int(b.get_ell_max(i)) for i in range(Nbins)]

r = results['r_unbinned']
Nsims = r.shape[1]

Cl_lens = b.bin_cell(hp.read_cl('../power_spectra/Cls_LiteBIRD_e2e_r0.fits')[2, :2*nside])[:Nbins]
Cl_tens = b.bin_cell(hp.read_cl('../power_spectra/Cls_Planck2018_tensor_r1.fits')[2, :2*nside])[:Nbins]

Cl_cmb = np.zeros((Nsims, Nbins))
for k in range(Nsims):
    Cl_cmb[k] = Cl_lens + r[:, k] * Cl_tens
```

```python
# Gerbino+20 Eq. (49a)

k = 0

def loglike_Cl(r, Cl_hat):
    Cl_th = Cl_lens + r* Cl_tens
    return (2*leff+1) * (Cl_hat / Cl_th - np.log(np.abs(Cl_hat/ Cl_th)) - 1)

N_grid = 1000
r_grid = np.linspace(0, 5e-2, N_grid)

loglike = np.zeros((Nbins, N_grid))
for i in range(N_grid):
    loglike[:, i] = loglike_Cl(r_grid[i], Cl_cmb[k])

# Plot for each multipole

for i in range(Nbins):
    plt.plot(r_grid, loglike[i])
plt.xlim(min(r_grid), max(r_grid))
#plt.xscale('log')
#plt.yscale('log')
plt.xlabel(r'$r$', fontsize=14)
plt.ylabel(r'$-2 \ln{\mathcal{L}(\mathcal{C}_\ell)}$', fontsize=14)
plt.show()
```

```python
# Gerbino+20 Eq. (51)

k = 0

def g(x):
    return np.sign(x-1) * np.sqrt(2 * (x-np.log(x)-1))

N_grid = 1000
r_grid = np.linspace(0, 1e-2, N_grid)

G = np.zeros((Nbins, N_grid))
for i in range(N_grid):
    Cl_th = Cl_lens + r_grid[i] * Cl_tens
    G[:, i] = g(Cl_cmb[k]/Cl_th) * Cl_lens

cov = np.cov(Cl_cmb.T)
invcov = np.linalg.inv(cov)

chi2 = 0
for i in range(2, Nbins):
    for j in range(2, Nbins):
        chi2 += G[i] * invcov[i,j] * G[j]

L = np.exp(-1/2 * chi2)

plt.plot(r_grid, L/max(L))
plt.xlim(min(r_grid), max(r_grid))
#plt.xscale('log')
#plt.yscale('log')
plt.xlabel(r'$r$', fontsize=14)
plt.ylabel('Relative probability', fontsize=14)
plt.show()
```

```python
# Offset Hamimeche-Lewis

k = 0
fsky_eff = 59.46942219524857 / 100

N_grid = 1000
r_grid = np.linspace(0, 1e-2, N_grid)

def compute_offset(ell, varcl, clref, fsky, N_iter=1):
    Nl = np.sqrt(np.abs(varcl - (2 / (2*ell+1) * clref**2) / fsky))
    for i in range(N_iter):
        Nl = np.sqrt(np.abs(varcl - 2 / (2*ell+1) / fsky * (clref**2 + 2 * Nl * clref)))
    return Nl * np.sqrt((2*ell+1) / 2)

def og(x):
    return np.sign(x) * g(np.abs(x))

offset = compute_offset(leff, np.diag(cov), Cl_lens, fsky_eff)
oCl_cmb = Cl_cmb + offset
oCl_lens = Cl_lens + offset

oG = np.zeros((Nbins, N_grid))
for i in range(N_grid):
    Cl_th = Cl_lens +  r_grid[i] * Cl_tens
    oCl_th = Cl_th + offset
    oG[:, i] = og(oCl_cmb[k]/oCl_th) * oCl_lens

cov = np.cov(Cl_cmb.T)
invcov = np.linalg.inv(cov)

chi2 = np.zeros(N_grid)
for i in range(Nbins):
    for j in range(Nbins):
        chi2 += oG[i] * invcov[i,j] * oG[j]

L = np.exp(-1/2 * chi2)

plt.plot(r_grid, L/max(L))
plt.xlim(min(r_grid), max(r_grid))
#plt.xscale('log')
#plt.yscale('log')
plt.xlabel(r'$r$', fontsize=14)
plt.ylabel('Relative probability', fontsize=14)
plt.show()
```

```python
# Final posterior

N_grid = 101
r_grid = np.linspace(0, 1e-2, N_grid)

chi2 = np.zeros((Nsims, N_grid))
L = np.zeros((Nsims, N_grid))
for k in range(Nsims):
    oG = np.zeros((Nbins, N_grid))
    for i in range(N_grid):
        Cl_th = Cl_lens +  r_grid[i] * Cl_tens
        oCl_th = Cl_th + offset
        oG[:, i] = og(oCl_cmb[k]/oCl_th) * oCl_lens

    for i in range(Nbins):
        for j in range(Nbins):
            chi2[k] += oG[i] * invcov[i,j] * oG[j]

    L[k] = np.exp(-1/2 * chi2[k])
    L[k] /= max(L[k])
    plt.plot(r_grid, L[k], color='black', alpha=0.1)

chi2_final = np.mean(chi2, axis=0)

posterior = np.exp(-1/2 * chi2_final)
posterior /= max(posterior)

plt.plot(r_grid, posterior, linewidth=2, color='red')
plt.xlim(min(r_grid), max(r_grid))
plt.xlabel(r'$r$', fontsize=14)
plt.ylabel('Relative probability', fontsize=14)
plt.show()
```

```python
first_moment = np.sum(r_grid * posterior) / np.sum(posterior)
second_moment = np.sum(r_grid**2 * posterior) / np.sum(posterior)

bias = r_grid[np.argmax(posterior)]
sigma_r = np.sqrt(second_moment - first_moment**2)

print('bias =', bias)
print('sigma_r =', sigma_r)
```

```python
# Marginalized posterior

N_grid_alpha = 51
alpha_grid = np.linspace(-2, 2, N_grid_alpha)
alpha_fid = 1

#tempFGRs = np.mean(Cl_cmb, axis=0) - Cl_lens

gnilc = np.load('../best_fits/results_d%ss%s_%s_scale%s_Nlbin%s_%s_gaussbeam_bandpass_ds_o%s%s_gnilc_fgres_v1_analytical.npy'
               % (dusttype, synctype, fsky, scale, Nlbin, cov_type, order, kw), allow_pickle=True).item()

noise = np.load('../best_fits/results_d%ss%s_%s_scale%s_Nlbin%s_%s_gaussbeam_bandpass_ds_o%s%s_gnilc_n_v1_analytical.npy'
               % (dusttype, synctype, fsky, scale, Nlbin, cov_type, order, kw), allow_pickle=True).item()

tempFGRs = np.mean(gnilc['r'].T * Cl_tens, axis=0) - np.mean(noise['r'].T * Cl_tens, axis=0)
tempFGRs = gnilc['r'].T * Cl_tens - np.mean(noise['r'].T * Cl_tens, axis=0)

#A = results['A_unbinned'].T / leff / (leff+1) * 2*np.pi
#tempFGRs = A / np.median(A) * np.median(sysFGRs)
"""
lbin = 28
ell_ini = np.concatenate((np.arange(2, lbin), np.arange(lbin, lmax+1, Nlbin)))
ell_end = np.concatenate((np.arange(2, lbin)+1, np.arange(lbin, lmax+1, Nlbin)+Nlbin))
ell_end = ell_end[ell_end <= lmax+1]
ell_ini = ell_ini[:len(ell_end)]
b_ini = nmt.NmtBin.from_edges(ell_ini, ell_end, is_Dell=True)
leff_ini = b_ini.get_effective_ells()

A = np.array([np.interp(leff, leff_ini, results['w1bw1b'][:, k]) for k in range(Nsims)]) / leff / (leff+1) * 2*np.pi
tempFGRs = A / np.median(A) * np.median(sysFGRs)
"""
#Nsims = len(tempFGRs)
chi2 = np.zeros((Nsims, N_grid, N_grid_alpha))
L = np.zeros((Nsims, N_grid, N_grid_alpha))
for k in range(Nsims):
    oG = np.zeros((Nbins, N_grid, N_grid_alpha))
    for i in range(N_grid):
        for j in range(N_grid_alpha):
            Cl_th = Cl_lens +  r_grid[i] * Cl_tens + alpha_grid[j] * np.mean(tempFGRs, axis=0)
            oCl_th = Cl_th + offset
            oCl_fid = Cl_lens + alpha_fid * np.mean(tempFGRs, axis=0) + offset
            oG[:, i, j] = og(oCl_cmb[k]/oCl_th) * oCl_fid

    for i in range(Nbins):
        for j in range(Nbins):
            chi2[k] += oG[i] * invcov[i,j] * oG[j]

chi2_final = np.mean(chi2, axis=0)
posterior2D = np.exp(-1/2 * chi2_final)
Nsims = r.shape[1]
```

```python
alpha_post = np.sum(posterior2D, axis=0) / np.max(np.sum(posterior2D, axis=0))
r_post = np.sum(posterior2D, axis=1) / np.max(np.sum(posterior2D, axis=1))

posterior = posterior[r_grid >= 0]
r_post = r_post[r_grid >= 0]
r_grid = r_grid[r_grid >= 0]

plt.plot(r_grid, posterior, '--', linewidth=2, color='red', label='before marg')
plt.plot(r_grid, r_post, linewidth=2, color='red', label='after marg')
plt.xlim(min(r_grid), max(r_grid))
plt.xlim(0, max(r_grid))
plt.xlabel(r'$r$', fontsize=14)
plt.ylabel('Relative probability', fontsize=14)
plt.legend(fontsize=12)
plt.show()

plt.plot(alpha_grid, alpha_post, linewidth=2, color='red')
plt.xlim(min(alpha_grid), max(alpha_grid))
plt.xlabel(r'$\alpha$', fontsize=14)
plt.ylabel('Relative probability', fontsize=14)
plt.show()
```

```python
first_moment = np.sum(r_grid * r_post) / np.sum(r_post)
second_moment = np.sum(r_grid**2 * r_post) / np.sum(r_post)

bias = r_grid[np.argmax(r_post)]
sigma_r = np.sqrt(second_moment - first_moment**2)
delta_r = r_grid[np.min(np.argwhere(np.cumsum(r_post) / np.sum(r_post) >= 0.68))]

print('bias =', bias)
print('delta_r =', delta_r)

peak_alpha = alpha_grid[np.argmax(alpha_post)]
print('\npeak alpha:', peak_alpha)
```
