```python
import sys
sys.path.append('../lib/')
import numpy as np
import healpy as hp
import pymaster as nmt
import matplotlib.pyplot as plt
from tqdm import trange
import covlib as cvl
```

```python
nside = 64
lmax = 2*nside-1
Nlbin = 1
fg_type = 'm'
Npix = hp.nside2npix(nside)

b = nmt.NmtBin.from_lmax_linear(lmax, nlb=Nlbin, is_Dell=False)
ell = b.get_effective_ells()
Nbins = len(ell)

Cl_lens = b.bin_cell(hp.read_cl('../power_spectra/Cls_LiteBIRD_e2e_r0.fits')[2, :2*nside])[:Nbins]
Cl_tens = b.bin_cell(hp.read_cl('../power_spectra/Cls_Planck2018_tensor_r1.fits')[2, :2*nside])[:Nbins]

results = np.load(f'../best_fits/results_d{fg_type}s{fg_type}_0.7_scale10_Nlbin10_Nmt-fg_gaussbeam_bandpass_ds_o1bts_adaptative_analytical.npy', allow_pickle=True).item()
Cl_cmb = results['cmb_unbinned'].T / ell / (ell+1) * 2*np.pi
N = len(Cl_cmb)

results_gnilc_fgres = np.load(f'../best_fits/results_d{fg_type}s{fg_type}_0.7_scale10_Nlbin10_Nmt-fg_gaussbeam_bandpass_ds_o1bts_adaptative_gnilc_fgres_v1_analytical.npy', allow_pickle=True).item()
Cl_gnilc_fgres = results_gnilc_fgres['cmb'].T / ell / (ell+1) * 2*np.pi
N_gnilc = len(Cl_gnilc_fgres)

results_gnilc_n = np.load(f'../best_fits/results_d{fg_type}s{fg_type}_0.7_scale10_Nlbin10_Nmt-fg_gaussbeam_bandpass_ds_o1bts_adaptative_gnilc_n_v1_analytical.npy', allow_pickle=True).item()
Cl_gnilc_n = results_gnilc_n['cmb'].T / ell / (ell+1) * 2*np.pi

Cl_temp = Cl_gnilc_fgres - Cl_gnilc_n
```

```python
alm_cmb = np.zeros((N, hp.Alm.getsize(lmax)), dtype=complex)
alm_temp = np.zeros((N_gnilc, hp.Alm.getsize(lmax)), dtype=complex)
alm_gnilc_fgres = np.zeros((N_gnilc, hp.Alm.getsize(lmax)), dtype=complex)
alm_gnilc_n = np.zeros((N_gnilc, hp.Alm.getsize(lmax)), dtype=complex)

for i in trange(N):
    for l in range(2, lmax+1):
        for m in range(l+1):
            idx = hp.Alm.getidx(lmax, l, m)
            alm_cmb[i, idx] = np.sqrt(Cl_cmb[i, l-2]+0j)

            if i < N_gnilc:
                alm_temp[i, idx] = np.sqrt(Cl_temp[i, l-2]+0j)
                alm_gnilc_fgres[i, idx] = np.sqrt(Cl_gnilc_fgres[i, l-2]+0j)
                alm_gnilc_n[i, idx] = np.sqrt(Cl_gnilc_n[i, l-2]+0j)
```

```python
cmb = np.zeros((N, Npix))
Cl_cmb2 = np.zeros_like(Cl_cmb)

gnilc_fgres = np.zeros((N_gnilc, Npix))
Cl_gnilc_fgres2 = np.zeros_like(Cl_gnilc_fgres)

gnilc_n = np.zeros((N_gnilc, Npix))
Cl_gnilc_n2 = np.zeros_like(Cl_gnilc_n)

temp = np.zeros((N_gnilc, Npix))
Cl_temp2 = np.zeros_like(Cl_temp)

for i in trange(N):
    cmb[i] = hp.alm2map(alm_cmb[i], nside)
    #cmb[i] = hp.synfast(np.concatenate((np.zeros(2), Cl_cmb[i])), nside)
    Cl_cmb2[i] = hp.anafast(cmb[i])[2:lmax+1]

    if i < N_gnilc:
        temp[i] = hp.alm2map(alm_temp[i], nside)
        #temp[i] = hp.synfast(np.concatenate((np.zeros(2), Cl_temp[i])), nside)
        Cl_temp2[i] = hp.anafast(temp[i])[2:lmax+1]

        gnilc_fgres[i] = hp.alm2map(alm_gnilc_fgres[i], nside)
        #gnilc_fgres[i] = hp.synfast(np.concatenate((np.zeros(2), Cl_gnilc_fgres[i])), nside)
        Cl_gnilc_fgres2[i] = hp.anafast(gnilc_fgres[i])[2:lmax+1]

        gnilc_n[i] = hp.alm2map(alm_gnilc_n[i], nside)
        #gnilc_n[i] = hp.synfast(np.concatenate((np.zeros(2), Cl_gnilc_n[i])), nside)
        Cl_gnilc_n2[i] = hp.anafast(gnilc_n[i])[2:lmax+1]
```

```python
hp.projview(np.mean(cmb, axis=0), cmap='planck', min=-0.1, max=0.1)
plt.show()
```

```python
tempFGRs = np.mean(Cl_gnilc_fgres2 - Cl_gnilc_n2, axis=0) #/ np.mean(Cl_gnilc_fgres2 - Cl_gnilc_n2) * np.mean(Cl_cmb2 - Cl_lens)
#tempFGRs =  np.mean(Cl_temp2, axis=0) #/ np.mean(Cl_temp2) * np.mean(Cl_cmb2 - Cl_lens)
#tempFGRs = np.mean(Cl_cmb, axis=0) - Cl_lens

plt.fill_between(ell, 1e-3*Cl_tens, 1e-2*Cl_tens, color='lightgrey', label=r'$r \in [10^{-3}, 10^{-2}]$')
plt.plot(ell, Cl_lens, ls=(0, (10,6)), lw=1, color='black', label='CMB lensing')
plt.plot(ell, np.mean(Cl_cmb, axis=0) - Cl_lens, label='fit')
#plt.plot(ell, np.mean(Cl_cmb2, axis=0) - Cl_lens, label='anafast')
#plt.plot(ell, np.mean(Cl_temp, axis=0), ls='--', lw=2, color='C0', label='temp fit')
plt.plot(ell, tempFGRs, ls='--', lw=2, color='C1', label='temp anafast')
plt.xlabel(r'$\ell$')
plt.ylabel(r'$C_\ell^{BB}$')
plt.xlim(min(ell), max(ell))
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.show()
```

```python
# Global fit no marg

total_cmb = Cl_cmb
cov_cmb = np.cov(total_cmb.T)
N_inv_cmb = np.linalg.inv(cov_cmb)
    
A_cmb = np.zeros((Nbins, 1))
A_cmb[:, 0] = Cl_tens
    
W_cmb = np.linalg.inv(A_cmb.T @ N_inv_cmb @ A_cmb) @ A_cmb.T @ N_inv_cmb
    
samp_nomarg = np.zeros((N, 2))
chi2r_nomarg = np.zeros(N)
    
for i in range(N):
    d_cmb = total_cmb[i] - Cl_lens
    s_cmb = W_cmb @ d_cmb
        
    res = d_cmb - A_cmb @ s_cmb
    dof = len(d_cmb) - len(s_cmb)
        
    samp_nomarg[i, 0] = s_cmb[0]
    chi2r_nomarg[i] = res.T @ N_inv_cmb @ res / dof
    
samp_nomarg[:, 1] = 100
    
r_global = np.mean(samp_nomarg[:, 0])
sigma_r_global = np.std(samp_nomarg[:, 0])

print(f'\nGlobal fit:\n'+
      f'r          = {np.round(r_global, 5)} +/- {np.round(sigma_r_global, 5)}\n'+
      f'bias/sigma = {np.abs(r_global/sigma_r_global)}'
     )
```

```python
# Global fit marg

A_cmb = np.zeros((Nbins, 2))
A_cmb[:, 0] = Cl_tens
A_cmb[:, 1] = tempFGRs
    
W_cmb = np.linalg.inv(A_cmb.T @ N_inv_cmb @ A_cmb) @ A_cmb.T @ N_inv_cmb
    
samp = np.zeros((N, 2))
chi2r = np.zeros(N)
    
for i in range(N):
    d_cmb = total_cmb[i] - Cl_lens
    s_cmb = W_cmb @ d_cmb
    
    res = d_cmb - A_cmb @ s_cmb
    dof = len(d_cmb) - len(s_cmb)
    
    samp[i] = s_cmb
    chi2r[i] = res.T @ N_inv_cmb @ res / dof
    
r_margin, alpha = np.mean(samp, axis=0)
sigma_r_margin, sigma_alpha = np.std(samp, axis=0)

print(f'\nMarginalized fit:\n'+
      f'r          = {np.round(r_margin, 5)} +/- {np.round(sigma_r_margin, 5)}\n'+
      f'bias/sigma = {np.abs(r_margin/sigma_r_margin)}\n'+
      f'alpha      = {np.round(alpha, 5)} +/- {np.round(sigma_alpha, 5)}'
     )
```

```python
nside = 64
lmax = 2*nside-1
fsky = '0.5_402_baseline'
Nlbin = 1
fg_type = 'baseline'
Npix = hp.nside2npix(nside)
Nfreqs = 22

results = np.load(f'../best_fits/results_d{fg_type[0]}s{fg_type[0]}_{fsky}_scale10_Nlbin10_Nmt-fg_gaussbeam_bandpass_ds_o1bts_adaptative_analytical.npy', allow_pickle=True).item()
Cl_cmb = results['cmb_unbinned'].T / ell / (ell+1) * 2*np.pi
N = len(Cl_cmb)

mask = hp.read_map('../masks/mask_fsky0.5_402_baseline_nside64_aposcale10.npy')

b = nmt.NmtBin.from_lmax_linear(lmax, nlb=Nlbin, is_Dell=False)
ell = b.get_effective_ells()
Nbins = len(ell)

Cl_lens = b.bin_cell(hp.read_cl('../power_spectra/Cls_LiteBIRD_e2e_r0.fits')[2, :2*nside])[:Nbins]
Cl_tens = b.bin_cell(hp.read_cl('../power_spectra/Cls_Planck2018_tensor_r1.fits')[2, :2*nside])[:Nbins]

f = nmt.NmtField(mask, None, spin=0, lmax=lmax)
w = nmt.NmtWorkspace()
w.compute_coupling_matrix(f, f, b)

path = f'/pscratch/sd/s/svinzl/B_modes_project/maps/maps_downgraded_nside{nside}_e2e_{fg_type}/'
W_ell = np.load('weights_baseline.npy')
Ncross = len(W_ell)




W_ell_ij = np.zeros((Nbins, Nfreqs, Nfreqs))
z = 0
for i in range(Nfreqs):
    for j in range(i, Nfreqs):
        W_ell_ij[:, i, j] = W_ell[z, :]
        W_ell_ij[:, j, i] = W_ell[z, :]
        z += 1




W_lm = np.zeros((Nfreqs, hp.Alm.getsize(lmax)), dtype=complex)

for l in range(2, lmax+1):
    for m in range(l+1):
        idx = hp.Alm.getidx(lmax, l, m)

        z = 0
        for i in range(Nfreqs):
            for j in range(i, Nfreqs):
                if i == j:
                    W_lm[i, idx] = np.sqrt(W_ell[z, l-2] + 0j)
                z += 1
```

```python
alm_cmb = np.zeros((N, hp.Alm.getsize(lmax)), dtype=complex)

for k in trange(N):
    maps = np.zeros((Nfreqs, 3, Npix))
    maps[:, 1:] = np.load(f'{path}{k}.npy')[0]
    alms = np.zeros((Nfreqs, hp.Alm.getsize(lmax)), dtype=complex)

    for i in range(Nfreqs):
        alms[i] = hp.map2alm(maps[i], lmax)[2]

    for i in range(hp.Alm.getsize(lmax)):
        alm_cmb[k, i] = np.sum(W_lm[:, i] * alms[:, i])
```

```python
Cl_cmb2 = np.zeros((N, Nbins))

for k in trange(N):
    f = nmt.NmtField(mask, [hp.alm2map(alm_cmb[k], nside, lmax)])
    cl_coupled = nmt.compute_coupled_cell(f, f)
    Cl_cmb2[k] = w.decouple_cell(cl_coupled)
```
